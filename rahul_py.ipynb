**leap year using function**

```python
def leap_year(n):
  for i in range(len(n)):

    yr=n[i]

    if(yr%4==0 and yr%100!=0) or (yr%100==0 and yr%400==0):
      print("True")
    else:
       print("false")

test_data = [1900, 2000, 2016, 1987]
leap_year(test_data)
```

**factorial of a number**

```python
def fact(num18):
  if num18<0:
   return (None)
  elif num18==0:
    return(1)
  else:
    mul=1
    for i in range(1,num18+1):
      mul*=i
    return(mul)

num18=int(input("enter a number to which we have to find factorial :"))
x=fact(num18)
print(x)
```
