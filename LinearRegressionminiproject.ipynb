# **Linear Regression**

**Find the best-fit line for data points.**

**Step 1.Create sample data:**

```python
import numpy as np
```

```python

def regression(z):            # defining function with name regression
  list=[]                     #taking a list to store variable
  for i in range(z):          # loop is taken to store value inside llits
    x1=int(input(f"please enter number {i+1} on  axis "))
    list.append(x1)
  return list

x=int(input("Please enter number of element on X axis :")) # function has been call for x axis
z5=regression(x)                                           # we have kept value of x axis list into z5 variable to convert into numpy array

y=int(input("Please enter number of element on y axis :"))  # function has been call for y axis
z6=regression(y)                                             # we have kept value of x axis list into z5 variable to convert into numpy array

```

**Step 2.converting a list into numpyArray**

```python
list_x=np.array(z5)             # element on x axis converted into numpy array
list_y=np.array(z6)             # element on y axis converted into numpy array
```

**We know that euation of best fit line is y=mx+b where m is slope and b is intercept**

# **Step 3
## Calculate slope (m) and intercept (b) using least squares:**

#   m = (mean(X×Y) - mean(X)×mean(Y)) / (mean(X²) - mean(X)²)
#  step 1 we calculating mean of X×Y

```python

#  step 1 we calculating mean of (X×Y)

mul_xy=list_x*list_y

mean_x=np.mean(list_x)      # calculating mean of x here
mean_y=np.mean(list_y)      # calculating mean of y here

mean_xy=np.mean(mul_xy)     # calculating mean of multiplication of xy
mulx_muly=mean_x*mean_y     # calculating  multiplication of mean of x and mean of y
```

calculating mean(X)²

```python
SquareofMean= np.square(mean_x) # here we calculated mean(X)² by squaring of mean of x
```

 calculating mean(X²)

```python
square_listx=np.square(list_x)  # here we calculated  square of x

mean0fsquare_listx=np.mean(square_listx)   #and then mean of that square list
```

**Calculating slope(m) using least squares**

```python
slope=(mean_xy-mulx_muly)/(mean0fsquare_listx-SquareofMean)
m=slope
```

**Calculating intercept (b) using least squares:**

```python
intercept=mean_y-m*mean_x
b=intercept
```

Printing equation of line

```python
print("Slope=",m)
print(" intercept=",b)
print(f"best fit line is y =({m})*X+{b}")
```
