**First step is togenerate a number betwwen 1 to 6**

```python
rolls=np.random.randint(1,7,100) # here we are using numpy function random.randint to print number between 1 to 6 100 times
```

**Print the generated number**

```python
print(f" 100 times random generated number between 1 to 6 is {rolls }")  #here we just printed the number 100 times as
                                                                          #we can  see and count number correctly
```

**total Count how many times you rolled a 6**

```python
y=np.sum(rolls==6)                   # here we taken a y variable and put the total no. of count of 6 appeared
# totalcount=print(counting)
totalcount=y
```
